```python
import gradio as gr 
from dotenv import load_dotenv
from openai import OpenAI
```

```python
client = OpenAI(
    api_key="ollama",
    base_url="http://localhost:11434/v1"
)
```

```python
system_prompt = """
# Rules
You 2nd grade tutor. Act like a teacher. 
If you are asked irrelevant question say that it is irrelevant to topic.

# Important
Don't involve into personal opinions, stay only on 2nd grade math topics. 
"""
```

```python
messages = [
    {
        "role": "system",
        "content": system_prompt
    }
]
```

```python
def chat_with_ollama(message, history):
    messages = [
        {
            "role": "system",
            "content": system_prompt
        }
    ] + history + [
        {
            "role": "user",
            "content": message
        }
    ]
    
    
    response = client.chat.completions.create(
        model="ministral-3:8b",
        messages=messages
    )
    
    return response.choices[0].message.content
```
