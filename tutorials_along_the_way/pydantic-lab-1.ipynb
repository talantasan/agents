```python
from pydantic import BaseModel, Field
```

```python
class Item(BaseModel):
    name: str = Field(min_length=1)
    price: float = Field(gt=0)
    quantity: int = Field(default=1, ge=1)
```

```python
def subtotal(item: Item) -> float:
    return item.price * item.quantity
```

```python
order_notebook = Item(name="Notebook", price=500.4, quantity=4)
order_keyboard = Item(name="Keyboard", price=23, quantity=4)
order_mouse = Item(name="Mouse", price=12.2, quantity=4)
print(subtotal(order_notebook))
print(subtotal(order_keyboard))
print(subtotal(order_mouse))
```
