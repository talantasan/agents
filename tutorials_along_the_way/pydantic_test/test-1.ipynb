```python
from pydantic import BaseModel, Field, ValidationError
from datetime import datetime, UTC
```

```python
class UserInfo(BaseModel):
    username: str
    email: str
    roles: list[str]
```

```python
user1 = UserInfo(username="Talant", email="talant@test.com", roles=["editor", "admin"])
```

```python
try:
    user2 = UserInfo(
        username="Jasmine",
        email="jasmin@example.kg",
        roles=["123", "125"]
    )
    output = user2.model_dump_json()
    print(output)
except ValidationError as e:
    print(e)
    
```

### Defining defaults in Pydantic

```python
class BlogPost(BaseModel):
    title: str
    content: str
    view_count: int = 0
    is_published: bool = False
    
    tags: list[str] = Field(default_factory=list)
    created_at: datetime = Field(default_factory=lambda: datetime.now(tz=UTC))
```

```python
blog1 = BlogPost(
    title="my post",
    content="This is my test post",
    tags=['12', 'best']
)
```

```python
print(blog1.model_dump_json(indent=2))
```
