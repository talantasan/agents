```python
from pydantic import BaseModel, Field, EmailStr, SecretStr, ValidationError
from datetime import datetime, UTC
from typing import Literal, Annotated
from uuid import UUID, uuid4
```

```python
class UserInfo(BaseModel):
    uid: UUID = Field(default_factory=uuid4)
    username: Annotated[str, Field(min_length=3, max_length=50)]
    password: SecretStr
    email: EmailStr
    roles: list[str]
```

```python
user1 = UserInfo(
    username="Talant", 
    email="talant@test.com", 
    password="myluv",
    roles=["editor", "admin"])
```

```python
try:
    user2 = UserInfo(
        username="Jasmine",
        email="jasmin@example.kg",
        password="myluv123",
        roles=["123", "125"]
    )
    output = user2.model_dump_json(indent=2)
    print(output)
except ValidationError as e:
    print(e)
    
```

### Defining defaults in Pydantic

```python
class BlogPost(BaseModel):
    title_id: Annotated[int, Field(gt=0)]
    content: Annotated[str, Field(min_length=5, max_length=100)]
    view_count: int = 0
    is_published: bool = False
    
    status: Literal["draft", "published", "archive"] = "draft"
    
    tags: list[str] = Field(default_factory=list)
    created_at: datetime = Field(default_factory=lambda: datetime.now(tz=UTC))
```

```python
blog1 = BlogPost(
    title_id=12,
    content="This is my test post",
    tags=['12', 'best'],
    status="published"
)
```

```python
print(blog1.model_dump_json(indent=2))
```
