```python
from openai import OpenAI
from pydantic import BaseModel, Field, ValidationError
from IPython.display import display, Markdown
```

```python
MODEL_ID="gemma4:e4b"

client = OpenAI(
    api_key="ollama",
    base_url="http://localhost:11434/v1"
)
```

```python
def chat_with_ollama(message: str) -> str:
    messages = [
        {"role": "system", "content": "You are a helpful assistant."},
        {"role": "user", "content": message}
    ]
    response = client.chat.completions.create(
        model=MODEL_ID,
        messages=messages
    )
    reply = response.choices[0].message.content
    return reply
```

```python
class GetItemDetailsArgs(BaseModel):
    """Item details to be retrieved"""
    name: str = Field(description="Name of the item, e.g keyboard")
# print(GetItemDetailsArgs.model_json_schema())

class ChangeCurrency(BaseModel):
    from_currency: str = Field(description="Source currency, e.g USD")
    to_currency: str = Field(description="")
```

```python
def get_item(args: GetItemDetailsArgs) -> str:
    fake_items = {
        "keyboard": 14.5,
        "monitor": 150, 
        "mouse": 20,
        "usb-c cable": 7.3, 
    }
    return {"Item": args.name, "Price": fake_items[args.name]}
```

```python
try:
    item1 = GetItemDetailsArgs(name="mouse")
    result = get_item(item1)
    print(result)
except ValidationError as e:
    print(e)
```
