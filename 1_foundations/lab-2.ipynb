````python
import gradio as gr
import json
from dotenv import load_dotenv
from openai import OpenAI
from pypdf import PdfReader
from IPython.display import display, Markdown

````

````python
load_dotenv(override=True)
client = OpenAI()
````

````python
reader = PdfReader("twin/linkedin.pdf")
linkedin_profile_pdf = ""
for page in reader.pages:
    text = page.extract_text()
    if text:
        linkedin_profile_pdf += text
print(display(Markdown(linkedin_profile_pdf)))
````

````python
with open('twin/summary.txt', 'r', encoding='utf-8') as f:
    linkedin_summary = f.read()
print(display(Markdown(linkedin_summary)))
````

````python
messages = [
    {
        "role": "system",
        "content": "Hi you are a helpful assistan!"
    },
    {
        "role": "user",
        "content": "My name is Talant"
    }
]
````

````python
response = client.chat.completions.create(
    model="gpt-5.4-nano",
    messages=messages
)
print(display(Markdown(response.choices[0].message.content)))
````

````python
system_prompt = f"""
# Your role

You are a digital twin running on a website, chatting with visitors of the website. 
You represent the person who's website you are on. 
You answer questions related to their career, background, skills and experience.

Here are the details of person you are representing:

{summary}

If asked, you explain clearly that you are an AI that is the digital twin of the person.

# Context

Here is the summary of the person's linkedin profile so that you can answer questions:

{linkedin_profile_pdf}


# Rules

Engage with the user. Be professional and engaging, as if talking to a potential client or future employer came across your profile.
Avoid answering questions that are not related to the person's career, background, skills, or experience.
Steer the conversation back to the professional topics.

Always stay in character as the digital twin of the person. Represent them accurately and professionally in all interactions.

IMPORTANT: If you don't know the answer, say no. Never make up an answer. 
If the user asks about something not in the context say that you don't know.
"""
````

````python
display(Markdown(system_prompt))
````

````python
messages = [
    {"role": "system", "content": system_prompt},
    {"role": "user", "content": "Hi - Tell about yourself"}
]
````

````python
response = client.chat.completions.create(
    model="gpt-5.4-nano",
    messages=messages
)

display(Markdown(response.choices[0].message.content))
````

````python
def chat(message, history):
    messages = [{"role": "system", "content": system_prompt}] + history + [{"role": "user", "content": message}]
    response = client.chat.completions.create(
        model="gpt-5.4-nano",
        messages=messages
    )
    return response.choices[0].message.content
````

````python
display(Markdown(chat("Hi - Tell about yourself, keep it as short as possible", [])))
````
