```python
import os
from IPython.display import display, Markdown
from dotenv import load_dotenv
from openai import OpenAI
```

```python
load_dotenv(override=True)
GEMINI_BASE_URL = "https://generativelanguage.googleapis.com/v1beta/openai/"
```

```python
if gemini_api_key:
    print(f"API key found: {gemini_api_key[:8]}...")
else:
    print("API key not found")
```

```python
client = OpenAI(
    api_key=gemini_api_key,
    base_url=GEMINI_BASE_URL
)
```

```python
question = """
Provide a 2nd grade math question to assess the student's understanding of basic math concepts.
Only one question should be provided.
The question should be suitable for a 2nd grade student.
The question should involve basic math operations like addition or subtraction.
The question should have only one correct answer.
The question should be clear and easy to understand for a 2nd grade student.
The question should not include any complex words or concepts beyond the 2nd grade level.
The question should be solvable without requiring any additional context or information.
The question should encourage the student to think and apply their math skills.
The question should be engaging and interesting for a 2nd grade student.
The question should be appropriate for the student's age and cognitive level.
"""
```

```python
response = client.chat.completions.create(
    model = "gemini-3.5-flash-lite",
    messages = [
        {
            "role": "system",
            "content": "You are a professional math tutor."
        },
        {
            "role": "user",
            "content": question
        }
    ]
)
generated_question = response.choices[0].message.content
print(display(Markdown(generated_question)))
```
