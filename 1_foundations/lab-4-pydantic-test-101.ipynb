```python
from pydantic import BaseModel
from typing import Literal
from IPython.display import display, Markdown
```

```python
class Task(BaseModel):
    title: str
    priority: Literal["low", "medium", "high"]
    status_code: Literal[200, 400, 404, 500, "pending"]
```

```python
task_1 = Task(
    title="Finish report",
    priority="high",
    status_code=200 
)
```

```python
print(task_1)
```
