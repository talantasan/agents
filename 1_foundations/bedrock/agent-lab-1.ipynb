```python
import boto3
from botocore.exceptions import ClientError
from IPython.display import display, Markdown
```

```python
bedrock_runtime = boto3.client('bedrock-runtime', region_name='us-east-1')
```

```python
model_id = "amazon.nova-lite-v1:0"
prompt ="Explain Bedrock service in AWS, precisely and concisely. Only with 3 bullet points."
```

```python
try:
    response = bedrock_runtime.converse(
        modelId=model_id,
        system=[{"text": "You are consice AWS cloud architect."}],
        messages=[
            {
                "role": "user",
                "content": [{"text": prompt}]
            }
        ],
        inferenceConfig={
            "maxTokens": 100,
            "temperature": 0.7
        }
    )
    output_blocks = response["output"]["message"]["content"]
    text_output = ("\n").join(block["text"] for block in output_blocks)
    usage = response.get("usage", {})
    if usage:
        display(Markdown("**Usage:**"))
        display(Markdown(str(usage)))
    display(Markdown("**Response:**"))
    display(Markdown(text_output))
except ClientError as e:
    display(Markdown(f"**Error:** {e}"))
```
