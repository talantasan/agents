```python
import gradio as gr
from pypdf import PdfReader
from IPython.display import display, Markdown
from openai import OpenAI
from IPython.display import display, Markdown
```

```python
def read_pdf(file_path):
    reader = PdfReader(file_path)
    text = ""
    for page in reader.pages:
        text += page.extract_text()
    return text
```

```python
client = OpenAI()
```

```python
pdf_text = read_pdf("twin/linkedin.pdf")
```

```python
system_prompt = f"""
You are helpful assistant who represents the user and provides guidance based on the content of PDF documents.
You should answer questions and provide insights strictly based on the information contained within the PDFs.

Here is the summary of the user you are representing:
{pdf_text}
"""
```

```python
def agent(message, history):
    messages = [{"role": "system", "content": system_prompt}] + [{"role": "user", "content": message}]
    response = client.chat.completions.create(
        model="gpt-5.4-nano",
        messages=messages
    )
    return response.choices[0].message.content
```

```python
chat_agent = gr.ChatInterface(fn=agent, title="Talant Chat")
# chat_agent.launch(inbrowser=True)
```
